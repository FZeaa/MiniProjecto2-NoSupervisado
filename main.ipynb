```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargar el archivo Excel
df = pd.read_excel('_94f47dfe96414485b5def55bccbd7cb8_Datos_textosODS.xlsx')

# 2. Configurar el tamaño del gráfico
plt.figure(figsize=(10, 6))

# 3. Crear un gráfico para contar cuántos textos hay por cada ODS
sns.countplot(data=df, x='ODS', palette='viridis')

# 4. Añadir detalles
plt.title('Distribución de textos por ODS', fontsize=14)
plt.xlabel('Número de ODS', fontsize=12)
plt.ylabel('Cantidad de textos (Frecuencia)', fontsize=12)

# 5. Mostrar el gráfico
plt.show()
```
